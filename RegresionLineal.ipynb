## Regresión lineal ##

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Creamos el conjunto de datos
df = pd.DataFrame({
    "horas":[1,2,3,4,5],
     "nota":[2,3,5,4,6]
})
x = df[["horas"]]
y = df[["nota"]]
print (x)
```

```python
#Creamos nuestro modleo
modelo = LinearRegression()
modelo.fit(x, y)
```

```python
#Imprimimos el interceptor y el coef
print("Interceptor: ", modelo.intercept_)
print("Pendiente: ", modelo.coef_[0])
```

```python
#Valores predichos
y_pred = modelo.predict
```

```python
#Realizamos la nueva prediccion 7
horas=float(input("Ingrese las horas de estudio"))
prediccion = modelo.predict([[horas]])
print("Prediccion:", prediccion)
```

Graficar con matplotlib el modelo y verificar si el nuevo valor corresponde a la predicción en la recta

```python
# Recalculamos las predicciones para asegurar que no haya errores
y_pred = modelo.predict(x)

# Dibujamos la línea y los puntos
plt.plot(x, y_pred, color='red')
plt.scatter(x, y, color='blue')
plt.scatter(x, y_pred, color='orange')
plt.scatter([[2.5]], modelo.predict([[2.5]]), color='green', s=150)

# Forzamos los ejes a mostrar solo números enteros
plt.xticks([1, 2, 3, 4, 5])
plt.yticks([2, 3, 4, 5, 6])

plt.show()
```
